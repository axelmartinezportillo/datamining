```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url="https://raw.githubusercontent.com/axelmartinezportillo/datamining/refs/heads/main/archivos/Ejercicio%201.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Seleccionamos solo los números
df_numeric = df.select_dtypes(include=['number'])

# Calculamos la matriz
matriz = df_numeric.corr()

# Mostramos la matriz
matriz
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(matriz, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de Correlación - Análisis de Calefacción")
plt.show()
```
